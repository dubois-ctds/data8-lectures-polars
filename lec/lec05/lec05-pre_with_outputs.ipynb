## Create a table from columns

```python
# datascience
streets = make_array('Bancroft', 'Durant', 'Channing', 'Haste')
streets
```

```python
# polars
streets_pl = pl.Series('values', ['Bancroft', 'Durant', 'Channing', 'Haste'])
streets_pl
```

```python
# datascience
Table()
```

```python
# polars
pl.DataFrame()
```

```python
# datascience
southside = Table().with_column('Streets', streets)
southside
```

```python
# polars
southside_pl = pl.DataFrame([
    (streets_pl).alias('Streets')
])
southside_pl
```

```python
# datascience
southside.with_column('Blocks from campus', np.arange(4))
```

```python
# polars
southside_pl.with_columns((pl.int_range(4, eager=True)).alias('Blocks from campus'))
```

```python
# datascience
Table().with_columns(
    'Streets', streets,
    'Blocks from campus', np.arange(4)
)
```

```python
# polars
pl.DataFrame([
    (streets_pl).alias('Streets'),
    (pl.int_range(4, eager=True)).alias('Blocks from campus')
])
```
