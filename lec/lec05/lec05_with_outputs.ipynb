## Columns of Tables are Arrays ##

```python
# datascience
nba_2026 = Table.read_table('nba_salaries.csv').where('season', 2026)
warriors = nba_2026.where('team', 'Golden State Warriors')
```

```python
# polars
nba_2026_pl = pl.read_csv('nba_salaries.csv').filter((pl.col('season') == (2026)))
warriors_pl = nba_2026_pl.filter((pl.col('team') == ('Golden State Warriors')))
```

```python
warriors
```

```python
# datascience
warriors.select('salary')
```

```python
# polars
warriors_pl.select('salary')
```

```python
# datascience
warriors.column('salary')
```

```python
# polars
warriors_pl.get_column('salary')
```

## Create a table from columns

```python
# datascience
streets = make_array('Bancroft', 'Durant', 'Channing', 'Haste')
streets
```

```python
# polars
streets_pl = pl.Series('values', ['Bancroft', 'Durant', 'Channing', 'Haste'])
streets_pl
```

```python
# datascience
Table()
```

```python
# polars
pl.DataFrame()
```

```python
# datascience
southside = Table().with_column('Street', streets)
southside
```

```python
# polars
southside_pl = pl.DataFrame([
    (streets_pl).alias('Street')
])
southside_pl
```

```python
# datascience
southside.with_column('Blocks from campus', np.arange(4))
```

```python
# polars
southside_pl.with_columns((pl.int_range(4, eager=True)).alias('Blocks from campus'))
```

```python
# datascience
Table().with_columns(
    'Street', streets,
    'Blocks from campus', np.arange(4)
)
```

```python
# polars
pl.DataFrame([
    (streets_pl).alias('Street'),
    (pl.int_range(4, eager=True)).alias('Blocks from campus')
])
```

```python
# datascience
# how might I get "Channing" from this table?
southside = southside.with_column('Blocks from campus', np.arange(4))
southside.where("Street", "Channing").column("Street").item(0)
```

```python
# polars
# how might I get "Channing" from this table?
southside_pl = southside_pl.with_columns((pl.int_range(4, eager=True)).alias('Blocks from campus'))
southside_pl.filter((pl.col("Street") == ("Channing"))).get_column("Street").item(0)
```

## W.E.B. Du Bois was a data scientist!

[W.E.B. Du Bois - Wikipedia](https://en.wikipedia.org/wiki/W._E._B._Du_Bois)

[Data from Library of Congress](https://www.loc.gov/pictures/collection/anedub/item/2013650354/)

```python
# datascience
du_bois = Table.read_table('du_bois.csv')
du_bois
```

```python
# polars
du_bois_pl = pl.read_csv('du_bois.csv')
du_bois_pl
```

```python
# datascience
du_bois.column('ACTUAL AVERAGE')
```

```python
# polars
du_bois_pl.get_column('ACTUAL AVERAGE')
```

```python
# datascience
du_bois.column('FOOD')
```

```python
# polars
du_bois_pl.get_column('FOOD')
```

```python
# datascience
# how to get the average amount spent on food (by income bracket)?
du_bois.column('ACTUAL AVERAGE') * du_bois.column('FOOD')
```

```python
# polars
# how to get the average amount spent on food (by income bracket)?
du_bois_pl.get_column('ACTUAL AVERAGE') * du_bois_pl.get_column('FOOD')
```

```python
# datascience
# how to include this amount in table?
food_dollars = du_bois.column('ACTUAL AVERAGE') * du_bois.column('FOOD')
du_bois.with_columns('Food $', food_dollars)
```

```python
# polars
# how to include this amount in table?
food_dollars_pl = du_bois_pl.get_column('ACTUAL AVERAGE') * du_bois_pl.get_column('FOOD')
du_bois_pl.with_columns(pl.Series('Food $', food_dollars_pl))
```

```python
du_bois
```

```python
# datascience
# how to include this amount in table? revisited
du_bois = du_bois.with_columns('Food $', food_dollars)
du_bois
```

```python
# polars
# how to include this amount in table? revisited
du_bois_pl = du_bois_pl.with_columns((food_dollars_pl).alias('Food $'))
du_bois_pl
```

```python
# datascience
# how to reorder the columns?
du_bois = du_bois.select('CLASS', 'ACTUAL AVERAGE', 'FOOD', 'Food $')
# alternatively..
# du_bois = du_bois.select(0, 1, 2, 3, 8, 4, 5, 6, 7)
du_bois
```

```python
# polars
# how to reorder the columns?
du_bois_pl = du_bois_pl.select('CLASS', 'ACTUAL AVERAGE', 'FOOD', 'Food $')
# alternatively..
# du_bois = du_bois.select(0, 1, 2, 3, 8, 4, 5, 6, 7)
du_bois_pl
```
